# Task 1 — Create the Variables

```python
#Task 1 — Create the Variables

Customer_Name = 'Sarah Johnson'

Monthly_Subscription_Fee = 8500

Months_in_Company = 10

Calls_to_Customer_Care_This_Month = 6

Missed_Payments = 2

Total_Monthly_Data_Usage_GB = 18

Minimum_Expected_Usage_GB = 20

Premium_Subscriber = False

# Task 2 — Calculate Annual Subscription Value

Annual_Subscription = Monthly_Subscription_Fee * 12

print(Annual_Subscription)

# Task 3 — Update Missed Payments

Missed_Payments += 1

print(Missed_Payments)

# Task 4 — Check Data Usage

Below_expected_data_usage = Minimum_Expected_Usage_GB > Total_Monthly_Data_Usage_GB

print(Below_expected_data_usage)

# Task 5 — Determine if Customer Complaints Are High

High_complaints = Calls_to_Customer_Care_This_Month > 5

print(High_complaints)

# Task 6 — Determine if Payment History Is Poor

Poor_payment_history = Missed_Payments >= 3

print(Poor_payment_history)

# Task 7 — Predict Churn Risk

High_Churn_Risk = High_complaints and Poor_payment_history

print(High_Churn_Risk)

# Task 8 — Determine Customer Retention Eligibility

Retention_offer = High_Churn_Risk or Below_expected_data_usage

print(Retention_offer)

# Task 9 — Generate a Customer Churn Report

print(f'''====================================================
      CUSTOMER CHURN PREDICTION REPORT
====================================================

Customer Name: {Customer_Name}

Monthly Subscription: {Monthly_Subscription_Fee}

Estimated Annual Value: {Annual_Subscription}

Months with Company: {Months_in_Company}

Customer Care Calls: {Calls_to_Customer_Care_This_Month}

Missed Payments: {Missed_Payments}

Monthly Data Usage: {Total_Monthly_Data_Usage_GB}

Premium Subscriber: 

High Complaints: {High_complaints}

Poor Payment History: {Poor_payment_history}

High Churn Risk: {High_Churn_Risk}

Retention Offer Needed: {Retention_offer}

====================================================''')
```
